## Iteration 1: Audio Transcription with Wav2Vec2

**Model**: **Wav2Vec2** from Hugging Face.

**Task**: Load a pre-trained model and transcribe the audio file `audio.wav` to text using **Wav2Vec2**.

### Steps:
1. Load the pre-trained `Wav2Vec2` model and the processor.
2. Load the audio file `audio.wav`, resample it to 16,000 Hz if necessary, and prepare it for inference.
3. Generate the transcription using the model.
4. Save the transcription for future comparison.

```python
from transformers import Wav2Vec2Processor, TFWav2Vec2ForCTC
import librosa
import tensorflow as tf


# Load model and processor
model_name = "facebook/wav2vec2-base-960h"
processor = Wav2Vec2Processor.from_pretrained(model_name)
model = TFWav2Vec2ForCTC.from_pretrained(model_name)  # Conversión explícita

def load_audio(file_path):
    """Load and resample audio file"""
    speech, rate = librosa.load(file_path, sr=16000)
    return speech, rate

def transcribe_audio(speech, rate):
    """Transcribe audio to text"""
    input_values = processor(speech, return_tensors="tf", sampling_rate=rate).input_values
    logits = model(input_values).logits
    predicted_ids = tf.argmax(logits, axis=-1)
    return processor.batch_decode(predicted_ids)[0]

# Usage
speech, rate = load_audio("audio.wav")
transcription_wav = transcribe_audio(speech, rate)
print(transcription_wav)
```

## Iteration 2: Audio Transcription with Whisper

**Task**: Use the **Whisper** model to transcribe the audio file `audio.wav`.

### Steps:
1. Load the pre-trained **Whisper** model from Hugging Face.
2. Transcribe the audio file `audio.wav` using Whisper.
3. Save the transcription for future comparison.

```python
import librosa
from transformers import WhisperProcessor, TFWhisperForConditionalGeneration

# Load the pre-trained Whisper processor and model
model_name = "openai/whisper-base"
processor = WhisperProcessor.from_pretrained(model_name)
model = TFWhisperForConditionalGeneration.from_pretrained(model_name)

def load_audio(file_path):
    """
    Function to load an audio file and adjust it to the required sampling rate.
    """
    audio, sample_rate = librosa.load(file_path, sr=16000)
    return audio, sample_rate

def transcribe_text(audio, rate):
    """
    Function to transcribe audio to text using the Whisper model.
    """
    input_features = processor(audio, return_tensors="tf", sampling_rate=rate).input_features
    
    generated_ids = model.generate(input_features)
    
    transcription = processor.batch_decode(generated_ids, skip_special_tokens = True)[0]
    return transcription

speech, rate = load_audio("audio.wav")
transcription_wh = transcribe_text(speech, rate)
print(transcription_wh)
```

## Iteration 5: Comparing the Architectures and Technologies of Wav2Vec2 and Whisper

**Task**: Explore and compare the underlying technologies and architectures of **Wav2Vec2** and **Whisper**, focusing on their respective approaches to automatic speech recognition (**ASR**).

### Steps:

1. **Understanding the Core Architectures**:
   - **Wav2Vec2**:
     - Explore the **sequence-to-sequence architecture** with an **encoder-decoder** model. This model uses attention mechanisms to align audio frames with transcribed text.
     - Investigate how the encoder processes audio features and how the decoder generates the transcription, focusing on the importance of the attention mechanism in mapping the audio input to text.
   - **Whisper**:
     - Analyze Whisper’s **Transformer-based architecture**, which is designed to handle long sequences of audio through **self-attention** mechanisms. Explore how Whisper leverages the Transformer’s ability to capture long-range dependencies, making it more robust in noisy environments and multilingual tasks.

2. **Comparing ASR Approaches**:
   - **Wav2Vec2**:
     - Learn about the **encoder** in Wav2Vec2, which converts the raw audio into a higher-dimensional representation, and the **decoder**, which generates text.
     - Explore how Wav2Vec2's approach to **attention** differs from Whisper’s, and how it impacts its ability to process audio in various contexts, such as clean vs. noisy environments.
   - **Whisper**:
     - Dive into the **self-attention mechanism** used in Whisper, and how it allows the model to focus on different parts of the input audio, improving transcription accuracy in complex audio scenarios (e.g., accents, background noise).
     - Explore Whisper’s **multitask learning** capabilities, where it can handle various tasks like language identification, speech translation, and transcription in a single model.

3. **Strengths and Weaknesses**:
   - **Wav2Vec2**:
     - Strength: Highly optimized for specific speech recognition tasks and languages, often performing well in clean audio environments.
     - Weakness: May struggle more with noise and accents due to the reliance on traditional sequence-to-sequence architecture.

### 1. Arquitecturas Principales

**Wav2Vec2** usa un modelo seq2seq (codificador-decodificador) con atención tradicional para alinear audio y texto. **Whisper** emplea un Transformer con autoatención, manejando mejor secuencias largas y ruido.

### 2. Enfoques de ASR

**Wav2Vec2** procesa audio en representaciones de alta dimensión y genera texto con atención clásica. **Whisper** usa autoatención para mayor precisión en audio complejo y es multitarea (transcripción, traducción, detección de idioma).

### 3. Fortalezas y Debilidades

- **Wav2Vec2:** Bueno en audio claro, pero sensible a ruido/acentos.

- **Whisper:** Robusto en entornos difíciles y multilingüe, pero más pesado computacionalmente.

### Conclusión

**Wav2Vec2** es eficiente para tareas específicas; **Whisper** es más versátil para escenarios complejos. Pero en reglas generales el **Whisper** funciona mejor.
